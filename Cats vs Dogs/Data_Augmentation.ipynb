```python
import keras
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img= image.load_img(r"Augmented data\00382-200124692.png", target_size=(256,256))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(img)
```

```python
# Initialize with augmentation parameters
datagen = ImageDataGenerator(
    # rescale=1./255,
    rotation_range=30,
    shear_range=0.2,
    zoom_range=0.2,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)
```

```python
type(img)
```

```python
img= image.img_to_array(img)
```

```python
type(img)
```

```python
img.shape
```

```python
input_batch= img.reshape(1,256,256,3)
```

```python
i=0
for output in datagen.flow(input_batch, batch_size=2, save_to_dir="Augmented data"):
    i+=1
    if i==10:
        break
```
